```python
!pip install PyPDF2 # Install the PyPDF2 library. Make sure to install the correct version.
from google.colab import drive
import PyPDF2

# Mount Google Drive
drive.mount('/content/drive')

# Download Data
data_dir = '/content/drive/MyDrive' # Input a data dir path from your mounted Google Drive

def extract_text_with_ey(pdf_path):
    ey_containing_text = []
    with open(pdf_path, 'rb') as pdf_file:
        pdf_reader = PyPDF2.PdfReader(pdf_file)
        num_pages = len(pdf_reader.pages)
        for page_num in range(num_pages):
            page = pdf_reader.pages[page_num]
            page_text = page.extract_text()
            for word in page_text.split():
                if "UBER TECHNOLOGIES, INC" in word.upper():
                    ey_containing_text.append(word)
    return ey_containing_text

# Example usage:
pdf_path = '/content/drive/MyDrive/RAG/data/10k/uber_10k_2023.pdf'  # Replace with the actual path to your PDF file in Google Drive.
extracted_text = extract_text_with_ey(pdf_path)

print(extracted_text) # To see the output, run the code.
```
